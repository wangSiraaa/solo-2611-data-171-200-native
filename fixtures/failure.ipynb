```python
threshold = 10
label = "original"
```

```python
import csv
from IPython.display import display, clear_output
rows=list(csv.DictReader(open("fixtures/input.csv")))
selected=[r for r in rows if int(r["value"]) >= threshold]
handle=display({"text/plain":"starting", "text/html":"<b>starting</b>"}, raw=True, display_id=True)
print("selected", len(selected))
```

```python
handle.update({"text/plain":label+": "+str(len(selected)), "text/html":"<b>"+label+": "+str(len(selected))+"</b>"},raw=True)
print("updated another cell display")
```
